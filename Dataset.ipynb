```python
from faker import Faker
from faker.providers import BaseProvider

# Custom provider for food product names
class FoodProvider(BaseProvider):
    def food_product(self):
        food_items = [
            "Apple Pie", "Banana Smoothie", "Cheddar Cheese", "Greek Yogurt", "Avocado Toast", "Chocolate Bar",
            "Sushi Roll", "Spaghetti Carbonara", "Chicken Nuggets", "Beef Steak", "Fish Tacos", "Mango Salsa",
            "Pasta Alfredo", "Caesar Salad", "Garlic Bread", "Orange Juice", "Granola Bar", "Vegetable Soup",
            "Shrimp Cocktail", "Blueberry Muffin", "Salmon Fillet", "Potato Chips", "Peanut Butter", "Ice Cream"
        ]
        return self.random_element(food_items)

# Initialize Faker and add the custom provider
fake = Faker()
fake.add_provider(FoodProvider)

# Generate random food product names
for _ in range(10):
    print(fake.food_product())
```
